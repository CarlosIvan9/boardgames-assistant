```python
from datetime import datetime

today = datetime.today().strftime("%Y%m%d")
print(today)
```

```python

# Memory of the chatbot will be reset every different day
# Specify an ID for the thread
config = {"configurable": {"thread_id": today}}
```

```python
input_message='How do you win at Memoir?'

answer = graph.invoke({"messages": [{"role": "user", "content": input_message}]}, config=config)
```
